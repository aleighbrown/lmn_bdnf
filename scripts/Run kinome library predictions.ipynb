# Reading in the sites and scoring with kinome library

```python
import kinase_library as kl
import pandas as pd
import glob
```

## Get all the CSV files we wrote out of R in run_kinome_library_plot_results.R - just p_value < 0.1

```python
all_times_up = glob.glob('/Users/annaleigh/Documents/GitHub/lmn_bdnf/data/*_psp_version2*')
all_times_up
```

## Loop through all these files, run the substrate prediction on each one, then write out the result for plotting and analyses in R

```python
for fle in all_times_up:
    this_file = pd.read_csv(fle,sep = '\t')

    this_file = this_file[this_file['p_value'] < 0.1]
    output_name = fle.replace("_for_enrichment_psp_version2", "_kinome_library_prediction")
    # make an empty data.frame
    results_df = pd.DataFrame(columns=['Score', 'Score Rank', 'Percentile', 'Percentile Rank', 'gene_names', 'phospho_position'])
    print(fle)
    for index, row in this_file.iterrows():
        try:
            s = kl.Substrate(row['uniprot_flank_larger'])
            tmp = s.predict()
            tmp['gene_names'] = row['gene_names']
            tmp['phospho_position'] = row['phospho_position']
            tmp.index.name = 'kinase'
            tmp.reset_index(inplace=True)
        except:
            print("NO")
            print(index)
            break
        
        # Concatenate instead of append (more efficient)
        results_df = pd.concat([results_df, tmp], ignore_index=True)
    
    print('Predictions done - writing')
    
    results_df.to_csv(output_name)




```
